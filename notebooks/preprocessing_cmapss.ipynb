### Censure $p_c \in \{0, 50, 70, 90\}$
### tirage train/test 80/20

```python
train_FD001 = pd.read_csv("CMAPSS/train_FD001_preprocessed.csv", index_col = 0)
train_FD004 = pd.read_csv("CMAPSS/train_FD004_preprocessed.csv", index_col = 0)
```

```python
l001 = random.sample(list(train_FD001.unit_number.unique()), 50) # tirage aléatoire sans remise
l004 = random.sample(list(train_FD004.unit_number.unique()), 50) # tirage aléatoire sans remise
```

```python
train_FD001_censored = train_FD001
train_FD001_censored['censored'] = 0
for i in l001:
    dtemp = train_FD001.loc[train_FD001.unit_number == i]
    k= random.sample(list(dtemp.time_cycle[0:-1]), 1) # on tire au hasard un cycle avant le dernier cycle pour censurer les données
    k = k[0]
    # print(k)
    train_FD001_censored = train_FD001_censored.loc[((train_FD001_censored.unit_number == i) & (train_FD001_censored.time_cycle <= k)) | (train_FD001_censored.unit_number != i)]
    train_FD001_censored = train_FD001_censored.reset_index(drop=True)
    train_FD001_censored.loc[train_FD001_censored.unit_number == i, 'censored'] = 1
```

```python
for i in range(1,14):
    train_FD001_censored['target_'+str(i*10)] = 0
    train_FD001_censored.loc[(train_FD001_censored.rul <= (10 * i)) & (train_FD001_censored.censored == 0), 'target_'+str(i*10)] = 1
    train_FD001_censored.loc[(train_FD001_censored.rul <= (10 * i)) & (train_FD001_censored.censored == 1), 'target_'+str(i*10)] = np.nan
```

```python
train_FD004_censored = train_FD004
train_FD004_censored['censored'] = 0
for i in l004:
    dtemp = train_FD004.loc[train_FD004.unit_number == i]
    k= random.sample(list(dtemp.time_cycle[0:-1]), 1) # on tire au hasard un cycle avant le dernier cycle pour censurer les données
    k = k[0]
    # print(k)
    train_FD004_censored = train_FD004_censored.loc[((train_FD004_censored.unit_number == i) & (train_FD004_censored.time_cycle <= k)) | (train_FD004_censored.unit_number != i)]
    train_FD004_censored = train_FD004_censored.reset_index(drop=True)
    train_FD004_censored.loc[train_FD004_censored.unit_number == i, 'censored'] = 1
```

```python
for i in range(1,14):
    train_FD004_censored['target_'+str(i*10)] = 0
    train_FD004_censored.loc[(train_FD004_censored.rul <= (10 * i)) & (train_FD004_censored.censored == 0), 'target_'+str(i*10)] = 1
    train_FD004_censored.loc[(train_FD004_censored.rul <= (10 * i)) & (train_FD004_censored.censored == 1), 'target_'+str(i*10)] = np.nan
```

```python
train_FD001_censored.to_csv("CMAPSS/train_FD004_censored.csv")
train_FD004_censored.to_csv("CMAPSS/train_FD001_censored.csv")
```
